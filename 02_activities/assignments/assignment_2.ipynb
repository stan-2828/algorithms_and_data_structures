  You are given a list containing `n` integers in the range `[0, n]`. Return a list of numbers that are missing from the range `[0, n]` of the array. If there is no missing number, return -1. Note, all the integers in the list may not be unique.
  
  ## Examples

  ### Example 1

  Input: `lst = [0, 2]`

  Output: [1]

  ### Example 2

  Input: `lst = [5, 0, 1]`

  Output: [2, 3, 4]

  ### Example 3

  Input: `lst = [6, 8, 2, 3, 5, 7, 0, 1, 10]`

  Output: [4, 9]

#### Starter Code for Question 3


```python
# This is an efficient way of implementing the solution
# Time O(n^2)

from typing import List

def missing_num(nums: List) -> int:
  # TODO

  miss_num = []

  # Find largest number in given list
  max_value = nums[0] # Set the initial max_value to the first element in the given list

  for number in nums:
    # Update the max_value if a larger number is found
    if number > max_value:
      max_value = number
  
  # Run through the numbers from 0 to max_value, find and append the missing numbers to a new list.
  in_seq = False # Flag to determine if number is in given list. Set to False initially

  for i in range(max_value+1): # Iterate through numbers from 0 to max_value
    for j in range(len(nums)): # Iterate through the given list
      if i == nums[j]:
        in_seq = True
        break
    # Determine if number is in list or not
    if in_seq == False: 
      miss_num.append(i)
    else:
      in_seq = False
  
  # Check if missing number list is empty
  if not miss_num:
    return -1
  
  return miss_num

```

```python
# This is a slightly more efficient method of implementing the solution
# Time O(n)

from typing import List

def missing_num_set(nums: List) -> int:
  # TODO
  miss_num = []
  nums_set = set()

  # Find largest number in given list
  max_value = nums[0] # Set the initial max_value to the first element in the given list

  for number in nums:
    nums_set.add(number)
    # Update the max_value if a larger number is found
    if number > max_value:
      max_value = number

  # Iterate through nums_set and find missing numbers
  for i in range(max_value+1):
    if i not in nums_set:
      miss_num.append(i)
  
  # Check if missing number list is empty
  if not miss_num:
    return -1
  
  return miss_num

```

```python
# Your answer here

from typing import List

def first_duplicate(nums: List[int]) -> int:
    seen = set()
    for num in nums:
        if num in seen:
            return num
        seen.add(num)
    return -1
```
