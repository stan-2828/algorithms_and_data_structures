````python
# Your answer here

from typing import List

# This implementation is more brute force and inefficient. It has time complexity of 0(n^2) 
# and space complexity of O(1)

def move_zeros_to_end(nums: List[int]) -> List[int]:
    # TODO

    for i in range(len(nums)-1, -1, -1):
        spaces_moved = 0
        for j in range(i, -1, -1):
            if nums[j] == 0 and spaces_moved != 0:
                for k in range(i-spaces_moved, i, 1):
                    nums[k] = nums[k+1]
                    nums[k+1] = 0
                spaces_moved = 0
                break
            elif nums[i] == 0 and spaces_moved == 0:
                break
            else:
                spaces_moved += 1
    return nums

    pass


# This is a more efficient implementation. Time and space complexity of O(n).

def move_zeros_to_end2(nums: List[int]) -> List[int]:
    new_list = []

    count = 0

    for i in range(len(nums)):
        if nums[i] != 0:
            new_list.append(nums[i])
            count += 1
    
    if count <= len(nums) - 1:
        for i in range(count, len(nums), 1):
            new_list.append(0)
    
    return new_list

# This is an even more efficient implementation with time complexity of O(n) and space 
# complexity of O(1)

def move_zeros_to_end3(nums: List[int]) -> List[int]:
    new_list_index = 0

    for i in range(len(nums)):
        if nums[i] != 0:
            nums[new_list_index] = nums[i]
            new_list_index += 1
    
    for i in range(new_list_index, len(nums), 1):
        nums[i] = 0

    return nums
````
